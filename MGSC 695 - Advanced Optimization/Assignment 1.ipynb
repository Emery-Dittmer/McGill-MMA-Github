# <font color=#FF0000>Problem 1</font>

Consider a small car manufacturing company that produces two products, carpets and safety films
(for glass). The company has three machines, A, B, and C, that can be used to produce the products.
The table below shows the processing time (in hours) required for each machine to produce one square
meter of each product.

|Machine |CARPETS| SAFETY FILMS|
|---|---|---|
|A| 3.5 |1.4|
|B |1.3 |2.0|
|C |2.0| 4.2|

The company has 200 hours of machine time available to produce carpets and 150 hours of machine
time available to produce safety films per week. The profit per square meter for carpets is 20 and for
safety films is 30. Formulate a linear programming model to determine how many square meters of each
product the company should produce to maximize their profit, subject to the machine time constraints.
Requirements:
• Create a python prototype using Gurobi to implement the linear programming model and obtain
the optimal solution of the given problem.
• Should the company consider increasing the maximum number of hours to manufacture carpets
and/or safety films? Why?

```python
import gurobipy as gb
from gurobipy import *
import numpy as np
import pandas as pd
```

```python
Machines=["A","B","C"]
Carpets=[3.5,1.3,2]
Saftey_Films=[1.4,2,4.2]
max_Carpets=200
max_Saftey_Films=150
Price_Carpets=20
Price_Saftey_Films=30

#create variables to numer of machines
machine_len=len(Machines)
```

```python
#setup problem
prob_1=gb.Model("Question 1")

#Create variables for carpets produced
Carp=prob_1.addVars(machine_len,lb=0,vtype=GRB.CONTINUOUS, name = ["Carpet Machine"+ str(i) for i in Machines])
Films=prob_1.addVars(machine_len,lb=0,vtype=GRB.CONTINUOUS, name = ["Film Machine"+ str(i) for i in Machines])

#set objective function 
prob_1.setObjective(sum(Price_Carpets*Carp[i]/Carpets[i] for i in range(machine_len))+
                    sum(Price_Saftey_Films*Films[i]/Saftey_Films[i] for i in range(machine_len)),
                    GRB.MAXIMIZE)


#set constraints

##setting the supply constrains for each supply centre
prob_1.addConstr(sum(Carp[i] for i in range(machine_len))<=max_Carpets)
prob_1.addConstr(sum(Films[i] for i in range(machine_len))<=max_Saftey_Films)

#optimize problem
prob_1.optimize()

for v in prob_1.getVars():
    print(v.varname,"=",v.x)

#print objective cost
print("\n")
print("The optimized Max value imum  is therefore ${:,}".format(round(prob_1.objVal)))
```

```python
print("The number of square meters of carpet is",200/1.3)
print("The number of square meters of film is",150/1.4)
```
